```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1, 2, 3, 4, 5])
Y = np.array([300, 500, 700, 900, 1100])
print(X)
print(Y)
```

```python
m = X.shape[0]
print(m)
```

```python
i = 0
X_i = X[i]
Y_i = Y[i]
print(X_i)
print(Y_i)
```

```python
plt.scatter(X, Y, marker='x', c='r')
plt.title("Housing Prices")
plt.xlabel("Price in thousands of dollars")
plt.ylabel("Size in 1000 square feet")
plt.show()
```

```python
w = 200
b = 100
print(w)
print(b)
```

```python
def model_output(X, w, b):
  m = X.shape[0]
  f_wb = np.zeros(m)
  for i in range(m):
    f_wb[i] = w * X[i] + b
  return f_wb
```

```python
temp_f_wb = model_output(X, w, b)
plt.plot(X, temp_f_wb, c='b', label = "Our Prediction")
plt.scatter(X, Y, marker = 'x', c = 'r', label = "Actual Values")
plt.title("Housing Prices")
plt.ylabel("Price in thousands of dollars")
plt.xlabel("Size in 1000 square feet")
plt.legend()
plt.show()
```
